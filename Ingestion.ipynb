# MedX Iterative Retrieval (offline)

Loads the JSONL corpus, builds a dependency-free BM25 index, and runs iterative retrieval with
keyword expansion (no API calls). For the Gemini-refined pipeline see `Gemini_RAG_Pipeline.ipynb`.

````python
from medx_retrieval import BM25Index, format_result, iterative_retrieve, load_jsonl

DATA_PATH = "Psichiatry_DSM-5.jsonl"
documents = load_jsonl(DATA_PATH)
index = BM25Index(documents)

len(documents)
````

````python
# Do not call CLI/argparse code from a notebook cell. Use this function API instead.
query = "criteria for major depressive disorder"
outcome = iterative_retrieve(index, query, rounds=3, top_k=5, refiner="keywords")

print("Stopped:", outcome.stop_reason)
for round_ in outcome.rounds:
    print(f"Round {round_.round_number}:", round_.queries)
````

````python
for result in outcome.sources:
    print(format_result(result))
    print("-" * 80)
````
